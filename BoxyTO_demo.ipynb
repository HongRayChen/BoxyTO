# BoxyTO researcher demo
Run from the repository root in an existing Torch/SciPy/PyVista environment. Select a JSON example, GPU, and new output folder below. Defaults run 400 epochs. See README.md for method, boundary definitions, feature controls, and validation.

```python
from pathlib import Path
from boxyto.optimize import optimize, read_config, render_run
from IPython.display import Image, display

config = read_config('examples/mbb_5axis.json')
config['optimization'].update(device='cuda:0', fe_backend='cupy')
output = Path('results/notebook_mbb')
config
```

Choose mbb/mount/pedal/link and 3axis/5axis in the filename. Other examples show random BCs, no-design regions, custom constraints, uniform feature scaling, and independent z scaling of fields 2 and 3. Edit configuration values directly; no notebook helper functions are required.

```python
optimize(config, output)
```
